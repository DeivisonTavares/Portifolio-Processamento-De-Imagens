```python
img = cv2.imread('valdivia(1).jpeg')
B, G, R, = cv2.split(img)

#Sub-plot do Azul(B)
plt.subplot(1,3,1)
plt.imshow(B,cmap='gray')
plt.axis('off')
plt.title('Azul(B)')

#Sub-plot do Verde(G)
plt.subplot(1,3,2)
plt.imshow(G,cmap='gray')
plt.axis('off')
plt.title('Verde(G)')

#Sub-plot do Vermelho(R)
plt.subplot(1,3,3)
plt.imshow(R,cmap='gray')
plt.axis('off')
plt.title('Vermelho(R)')
```
