```python
img = cv2.imread('valdivia.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img__hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb
```

```python
plt.figure(figsize=(10,10))
plt.imshow(img)
plt.title('Mago Valvidia BGR')
plt.show()
```

```python
fig, ax = plt.subplots(1,3,figsize=(8,5))
ax[0].imshow(img)
ax[0].set_title('Mago Valvidia BGR')

ax[1].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
ax[1].set_title('Mago Valvidia RGB')

ax[2].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2HSV))
ax[2].set_title('Mago Valvidia HSV')

plt.show()
```

```python
fig, ax = plt.subplots(2,2,figsize=(8,5))

ax[0,0].imshow(img)
ax[0,0].set_title('Mago Valvidia BGR')
ax[0,0].axis('off')

ax[0,1].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2GRAY), cmap= 'gray')
ax[0,1].set_title('Mago Valvidia BGR2GRAY')
ax[0,1].axis('off')

ax[1,0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
ax[1,0].set_title('Mago Valvidia RGB')
ax[1,0].axis('off')

ax[1,1].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2HSV))
ax[1,1].set_title('Mago Valvidia HSV')
ax[1,1].axis('off')

plt.tight_layout()
plt.savefig('conversao_cores.jpg')

plt.show()
```
