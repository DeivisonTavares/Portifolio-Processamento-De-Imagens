```python
from google.colab import files
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
```

```python
treecko_upload = files.upload()
```

```python
treecko_bgr = cv2.imread('treecko2.jpg')
```

```python
treecko_gray = cv2.cvtColor(treecko_bgr, cv2.COLOR_BGR2GRAY)
```

```python
cv2_imshow(treecko_gray)
```

```python
def calcularHist(img):
    imgVetor = img.ravel()
    histVetor = np.zeros(256,dtype=np.int64)
    for px in imgVetor:
        histVetor[px] += 1
    return histVetor
```

```python
calcularHist(treecko_gray)
```

```python
histImg = calcularHist(treecko_gray)
plt.bar(range(256),histImg)
plt.show
```
